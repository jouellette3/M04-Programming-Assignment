```python
#16.4
import sqlite3

dbase = sqlite3.connect('books.db')
cursor = dbase.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS books (
    title TEXT,
    author TEXT,
    year INTEGERY
)
''')

dbase.commit()
```

```python
%%writefile books2.csv
title,author,year
The Weirdstone of Brisingamen,Alan Garner,1960
Perdido Street Station,China Miéville,2000
Thud!,Terry Pratchett,2005
The Spellman Files,Lisa Lutz,2007
Small Gods,Terry Pratchett,1992
```

```python
#16.6
import csv

with open('books2.csv', 'r') as file:
    reader = csv.DictReader(file)

    for row in reader:
        cursor.execute(
            'INSERT INTO books (title, author, year) VALUES (?, ?, ?)',
            (row['title'], row['author'], row['year'])
        )

dbase.commit()
```

```python
import sqlalchemy as sa

engine = sa.create_engine('sqlite:///books.db')

with engine.connect() as conn:
    result = conn.execute(sa.text('SELECT title FROM books ORDER BY title'))

    for row in result:
        print(row[0])                        
```
